```python
import numpy as np

# Create subject names
subjects = np.array([
    "Python",
    "Java",
    "Mathematics",
    "Database",
    "Computer Networks"
])

# Generate marks for 50 students in 5 subjects
np.random.seed(42)

marks = np.random.randint(35, 100, size=(50, 5))

# Display marks
print("STUDENT MARKS:")
print(marks)

print("\nSUBJECTS:")
print(subjects)
```

```python
print("Array Shape:", marks.shape)
print("Number of Dimensions:", marks.ndim)
print("Total Elements:", marks.size)
print("Data Type:", marks.dtype)

# Calculate total and average marks
total_marks = np.sum(marks, axis=1)
average_marks = np.mean(marks, axis=1)

print("\nSTUDENT SUMMARY")
print("-------------------------")

for i in range(50):
    print(
        f"Student {i+1}: "
        f"Total = {total_marks[i]}, "
        f"Average = {average_marks[i]:.2f}"
    )
```

```python
# First 10 students
print("FIRST 10 STUDENTS:")
print(marks[:10])

# Last subject marks using negative indexing
print("\nLAST SUBJECT MARKS:")
print(marks[:, -1])

# First 5 students and first 3 subjects
print("\nFIRST 5 STUDENTS, FIRST 3 SUBJECTS:")
print(marks[:5, :3])

# Students 11 to 20
print("\nSTUDENTS 11 TO 20:")
print(marks[10:20, :])

# Last 5 students
print("\nLAST 5 STUDENTS:")
print(marks[-5:, :])
```

```python
# Students scoring below 40 in any subject
failed_marks = marks < 40

print("MARKS BELOW 40:")
print(marks[failed_marks])

# Count failed subjects
print("\nTOTAL FAILED SUBJECTS:", np.sum(failed_marks))

# Students scoring above 90
high_scores = marks > 90

print("\nMARKS ABOVE 90:")
print(marks[high_scores])

print("TOTAL MARKS ABOVE 90:", np.sum(high_scores))

# Grade bands based on individual subject marks
grade_A = marks >= 90
grade_B = (marks >= 75) & (marks < 90)
grade_C = (marks >= 60) & (marks < 75)
grade_D = (marks >= 40) & (marks < 60)
grade_F = marks < 40

print("\nGRADE-WISE ANALYSIS")
print("Grade A (90-100):", np.sum(grade_A))
print("Grade B (75-89):", np.sum(grade_B))
print("Grade C (60-74):", np.sum(grade_C))
print("Grade D (40-59):", np.sum(grade_D))
print("Grade F (Below 40):", np.sum(grade_F))
```

```python
print("\nSUBJECT-WISE PERFORMANCE")
print("=" * 50)

for i in range(len(subjects)):

    subject_marks = marks[:, i]

    # Highest marks
    highest = np.max(subject_marks)

    # Student who scored highest
    topper = np.argmax(subject_marks)

    # Average marks
    average = np.mean(subject_marks)

    # Count students who passed (40 or above)
    passed = np.sum(subject_marks >= 40)

    # Count students who failed
    failed = np.sum(subject_marks < 40)

    # Pass percentage
    pass_percentage = (passed / 50) * 100

    print("\nSubject:", subjects[i])
    print("Highest Marks:", highest)
    print("Topper: Student", topper + 1)
    print("Average Marks:", round(average, 2))
    print("Passed Students:", passed)
    print("Failed Students:", failed)
    print("Pass Percentage:", round(pass_percentage, 2), "%")
```

```python
# Find overall topper
topper_index = np.argmax(total_marks)

print("OVERALL CLASS RESULT")
print("=" * 40)

print("Overall Topper: Student", topper_index + 1)
print("Total Marks:", total_marks[topper_index])
print("Average Marks:", round(average_marks[topper_index], 2))

# Class average
class_average = np.mean(marks)

print("Class Average:", round(class_average, 2))

# Students who passed all subjects
passed_all = np.all(marks >= 40, axis=1)

print("Students Passed All Subjects:", np.sum(passed_all))

# Students who failed at least one subject
failed_students = np.any(marks < 40, axis=1)

print("Students Failed in At Least One Subject:", np.sum(failed_students))
```

```python
print("\n" + "=" * 55)
print("         STUDENT MARKS ANALYZER REPORT")
print("=" * 55)

print("Total Students:", marks.shape[0])
print("Total Subjects:", marks.shape[1])
print("Maximum Marks:", np.max(marks))
print("Minimum Marks:", np.min(marks))
print("Class Average:", round(np.mean(marks), 2))
print("Overall Topper: Student", np.argmax(total_marks) + 1)

print("\nSUBJECT-WISE AVERAGES")

for i in range(5):
    print(subjects[i], ":", round(np.mean(marks[:, i]), 2))

print("\nGRADE DISTRIBUTION")
print("A Grade:", np.sum(marks >= 90))
print("B Grade:", np.sum((marks >= 75) & (marks < 90)))
print("C Grade:", np.sum((marks >= 60) & (marks < 75)))
print("D Grade:", np.sum((marks >= 40) & (marks < 60)))
print("F Grade:", np.sum(marks < 40))

print("\n" + "=" * 55)
print("             END OF REPORT")
print("=" * 55)
```
